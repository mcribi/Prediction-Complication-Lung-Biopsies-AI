```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

# Cargar el volumen NIfTI comprimido (.nii.gz)
img = nib.load("/mnt/homeGPU/mcribilles/TFG/volumenes/nifti_convertidos_anonimizados/1MASh.nii.gz")
vol = img.get_fdata()

# Elegir el corte central
mid_slice = vol.shape[2] // 2
slice_central = vol[:, :, mid_slice]

# Crear la figura
plt.figure(figsize=(6,6))
plt.imshow(np.rot90(slice_central), cmap='gray')
plt.axis('off')

# Guardar como TIFF a 300 dpi (recomendado para publicaciones)
plt.savefig("corte_central.eps", bbox_inches='tight', pad_inches=0, dpi=300)

plt.close()

print("Guardado como corte_central.tiff (300 dpi)")
```
